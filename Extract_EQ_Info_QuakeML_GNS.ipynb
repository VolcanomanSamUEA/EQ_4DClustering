```python
import requests
from xml.etree import ElementTree as ET

def get_event_information(event_id):
    base_url = "https://service.geonet.org.nz/fdsnws/event/1/query"
    params = {
        "eventid": event_id,
    }

    headers = {
        "Accept": "application/xml",
    }

    try:
        response = requests.get(base_url, params=params, headers=headers)

        # Check if the request was successful (status code 200)
        if response.status_code == 200:
            # Parse the XML response
            root = ET.fromstring(response.content)

            # Define the namespace
            namespace = {'bed': 'http://quakeml.org/xmlns/bed/1.2'}

            # Iterate through event elements
            for event_element in root.findall(".//bed:event", namespaces=namespace):
                # Iterate through origin elements
                for origin_element in event_element.findall(".//bed:origin", namespaces=namespace):
                    latitude_value = origin_element.find(".//bed:latitude/bed:value", namespaces=namespace).text
                    latitude_uncertainty = origin_element.find(".//bed:latitude/bed:uncertainty", namespaces=namespace).text

                    longitude_value = origin_element.find(".//bed:longitude/bed:value", namespaces=namespace).text
                    longitude_uncertainty = origin_element.find(".//bed:longitude/bed:uncertainty", namespaces=namespace).text

                    depth_value = origin_element.find(".//bed:depth/bed:value", namespaces=namespace).text
                    depth_uncertainty = origin_element.find(".//bed:depth/bed:uncertainty", namespaces=namespace).text

                    # Print or process the extracted information
                    print(f"Latitude: {latitude_value}, Uncertainty: {latitude_uncertainty}")
                    print(f"Longitude: {longitude_value}, Uncertainty: {longitude_uncertainty}")
                    print(f"Depth: {depth_value}, Uncertainty: {depth_uncertainty}")

        else:
            print(f"Failed to retrieve data. Status code: {response.status_code}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage with a specific event ID
event_id_to_query = "2016p858000"
get_event_information(event_id_to_query)
```

```python
import requests
from xml.etree import ElementTree as ET
import pandas as pd
import time

def get_event_information(event_id):
    base_url = "https://service.geonet.org.nz/fdsnws/event/1/query"
    params = {
        "eventid": event_id,
    }

    headers = {
        "Accept": "application/xml",
    }

    try:
        response = requests.get(base_url, params=params, headers=headers)

        # Check if the request was successful (status code 200)
        if response.status_code == 200:
            # Parse the XML response
            root = ET.fromstring(response.content)

            # Define the namespace
            namespace = {'bed': 'http://quakeml.org/xmlns/bed/1.2'}

            # Extract latitude, longitude, and depth information with adjusted XPath expressions
            latitude_value = root.find(".//bed:latitude/bed:value", namespaces=namespace).text
            longitude_value = root.find(".//bed:longitude/bed:value", namespaces=namespace).text
            depth_value = root.find(".//bed:depth/bed:value", namespaces=namespace).text
            latitude_uncertainty = root.find(".//bed:latitude/bed:uncertainty", namespaces=namespace).text
            longitude_uncertainty = root.find(".//bed:longitude/bed:uncertainty", namespaces=namespace).text
            depth_uncertainty = root.find(".//bed:depth/bed:uncertainty", namespaces=namespace).text

            # Return the extracted information as a dictionary
            return {
                "EventID": event_id,
                "Latitude": latitude_value,
                "Longitude": longitude_value,
                "Depth": depth_value,
                "Latitude_Uncertainty": latitude_uncertainty,
                "Longitude_Uncertainty": longitude_uncertainty,
                "Depth_Uncertainty": depth_uncertainty,
            }

        else:
            print(f"Failed to retrieve data for event ID {event_id}. Status code: {response.status_code}")
            return None
    except Exception as e:
        print(f"An error occurred for event ID {event_id}: {e}")
        return None

# Create an empty list to store extracted information
extracted_info_list = []

# Iterate through all 'publicid' values in the DataFrame
for event_id in df['publicid']:
    event_info = get_event_information(event_id)

    if event_info:
        extracted_info_list.append(event_info)
        print(f"{event_id} added to event_info")

    # Introduce a sleep interval to avoid potential rate limiting
    time.sleep(1)

# Create a new DataFrame with the extracted information
extracted_info_df = pd.DataFrame(extracted_info_list)

# Display the new DataFrame
print(extracted_info_df)
```
